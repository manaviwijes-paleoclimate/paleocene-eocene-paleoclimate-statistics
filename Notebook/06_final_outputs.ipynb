### Segmented linear regression of the carbon isotope record

A three-segment piecewise-linear regression model was fitted to 247 bulk carbonate δ¹³C observations from ODP Site 1262 spanning 56.30–55.65 Ma. The model identified two boundaries, bracketed by observations at 55.932–55.905 Ma and 55.870–55.867 Ma.

The older segment (n = 125; 56.299–55.932 Ma) exhibited a fitted δ¹³C decline toward younger ages of −0.00233‰/kyr. The middle segment (n = 22; 55.905–55.870 Ma) showed a steeper negative trend of −0.00418‰/kyr. In contrast, the younger segment (n = 100; 55.867–55.650 Ma) exhibited an increasing δ¹³C trend of +0.00333‰/kyr.

The three-segment linear model achieved a residual sum of squares (RSS) of approximately 3.038, compared with 14.968 for the three-segment piecewise-constant model. Among the four candidate models evaluated, the three-segment linear model had the lowest exploratory BIC (−1042.32).

Sensitivity analysis showed that the first fitted boundary remained unchanged across minimum segment sizes of 10–30 observations, whereas the second boundary shifted toward younger ages when larger minimum segments were imposed.

Residual lag-1 autocorrelation remained positive in all three segments (0.799, 0.629 and 0.865, respectively), indicating substantial serial dependence not captured by the fitted linear trends. Furthermore, the first boundary coincided with a 27 kyr sampling gap, limiting temporal resolution of the associated isotopic transition.

Consequently, the fitted segments are interpreted as exploratory descriptions of the sampled δ¹³C trajectory rather than precise estimates of geological event boundaries or instantaneous rates of carbon isotope change.
